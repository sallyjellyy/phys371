```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
```

## Generate the data to plot

```python
# Define an analytical function that we want to plot
def f(x):
  return x**(-1)

# Generate an array of x values for plotting
XMIN = 0.1
XMAX = 5.0
NUM = 200
x_vals = np.linspace(start=XMIN, stop=XMAX, num=NUM, endpoint=True)

# Generate an array of y values for plotting
y_vals = f(x_vals)
```

## Make a nice figure

```python
# This command creates the figure object
plt.figure(figsize=[8, 5]) # Size is 8x5 inches

# Plot y(x) as a blue line
plt.plot(x_vals, y_vals, '-', color='blue')

# Plot a point
point1_x = 0.5
point1_y = 3.0
plt.plot([point1_x], [point1_y], '.', color='red', label='point 1')

# Plot a point with errorbars
point2_x = 2.0
point2_y = 1.3
err_x = 0.5
err_y = 1.0
plt.errorbar(x=[point2_x], y=[point2_y], xerr=[err_x], yerr=[err_y],
             color='red', label='point 2')

# Label the axes -- always make sure to include the correct units!
plt.xlabel('$x$ quantity name (units)')
plt.ylabel('$y$ quantity name (units)')
# The $ symbols allow you to typeset math expressions.
# Ex: $R_{?}$ has the '?' as a subscript

# Ajust the x and y limits
plt.xlim([0, np.max(x_vals)])
plt.ylim([0, np.max(y_vals)])

# Add a legend
plt.legend()

# Save the figure
# To save this figure, uncomment the code below
# (remove the '#' in front of each line).

# filename = 'lab1_plot.png'
# plt.tight_layout() # This just helps the formatting
# plt.savefig(filename)
```
